```python
gamma = 0.9
theta = 0.000001
V0 = 0.0
V1 = 0.0
iterations = 0
while True:

    new_V0 = -1 + 0.81 * V1
    new_V1 = 89.9 + 0.09 * V0
    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))
    V0, V1 = new_V0, new_V1
    iterations += 1
    if delta < theta:
        break

Q0_down = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
Q0_right = -1

Q1_right = 0.9 * 100 + 0.1 * (-1 + gamma * V0)
Q1_up = -1 + gamma * V0

print("ASSIGNMENT 2 RESULTS")
print("Number of iterations:", iterations)
print(f"V_pi(S0)       = {V0:.2f}")
print(f"V_pi(S1)       = {V1:.2f}")
print(f"Q_pi(S0, Down) = {Q0_down:.2f}")
print(f"Q_pi(S0, Right)= {Q0_right:.2f}")
print(f"Q_pi(S1, Right)= {Q1_right:.2f}")
print(f"Q_pi(S1, Up)   = {Q1_up:.2f}")

print("\nPolicy Consistency Checks:")
print("V(S0) = Q(S0, Down):", abs(V0 - Q0_down) < theta)
print("V(S1) = Q(S1, Right):", abs(V1 - Q1_right) < theta)
```
