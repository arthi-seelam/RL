```python
gamma = 0.9
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}
normal_reward = -1
goal_reward = 10
def state_value_function(start_state):
    path = {
        "S0": [(-1), (-1), (-1), (10)],
        "S1": [(-1), (-1), (10)],
        "S3": [(-1), (10)],
        "S4": [(10)]
    }

    rewards = path[start_state]

    value = 0

    for t, reward in enumerate(rewards):
        value += (gamma ** t) * reward

    return value
def action_value_function(state, action):

    if state == "S0" and action == "Up":
        reward = -1

        return reward + gamma * state_value_function("S0")
print("State Values:")

for state in ["S0", "S1", "S3", "S4"]:
    print(state, "=", state_value_function(state))

q_value = action_value_function("S0", "Up")

print("\nAction Value:")
print("Q(S0, Up) =", q_value)
```
